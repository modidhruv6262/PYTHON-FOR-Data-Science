# 6.
Write a Python program to find the first appearance of the substring 'not' and'poor' from a given string, if 'not' follows the 'poor', replace the whole 'not'...'poor' substring with 'good'. Return the resulting string

```python
text = 'The lyrics are not that poor!'

idx_not = text.find('not')
idx_poor = text.find('poor')
if idx_not < idx_poor:
    text = text[:idx_not] + 'good' + text[idx_poor + 4:]

print(text)
```

Write a Python program to check whether a list contains a sublist.

```python
main_list = [1, 2, 3, 4, 5]
sub_list = [2, 3, 4]

found = False
n = len(sub_list)

for i in range(len(main_list) - n + 1):
    if main_list[i : i + n] == sub_list:
        found = True
        break

print(found)
```

# 15.
Given a number n, write a python program to make and print the list of Fibonacci  
series up to n. Input : n=7 Hint : first 7 numbers in the series Expected output :
First few Fibonacci numbers are 0, 1, 1, 2, 3, 5, 8, 13

```python
n = 7
fib_list = [0, 1]

while True:
    next_val = fib_list[-1] + fib_list[-2]
    if next_val > n:
        break
    fib_list.append(next_val)

print(f"Fibonacci numbers up to value {n}: {fib_list}")
```

# 16.
Counting the frequencies in a list using a dictionary in Python. Input : [1, 1, 1, 5, 5, 3, 1, 3, 3, 1,4, 4, 4, 2, 2, 2, 2]  
Expected output : 1 : 5 , 2 : 4 , 3 : 3 , 4 : 3 , 5 : 2

```python
data = [1, 1, 1, 5, 5, 3, 1, 3, 3, 1, 4, 4, 4, 2, 2, 2, 2]
frequencies = {}

for num in data:
    if num in frequencies:
        frequencies[num] += 1
    else:
        frequencies[num] = 1
frequencies = dict(sorted(frequencies.items(), key=lambda keys: keys[0]))
print(frequencies)
```

# 17.
Write a python program using function to find the sum of odd series andevenseries  
Odd series: 12/ 1! +32/ 3! + 52/ 5!+……n  
Even series: 22/ 2! + 42/ 4! + 62/ 6!+……n   

```python
def get_factorial(num):
    fact = 1
    for i in range(1, num + 1):
        fact *= i
    return fact

def calculate_series(n, mode="odd"):
    total_sum = 0
    
    start = 1 if mode == "odd" else 2
    
    for x in range(start, n + 1, 2):
        term = (x**2) / get_factorial(x)
        total_sum += term
        print(f"Term {x}: {x}^2 / {x}! = {term}")
        
    return total_sum

n_limit = 6

odd_result = calculate_series(n_limit, "odd")
even_result = calculate_series(n_limit, "even")

print(f"\nFinal Odd Sum: {odd_result}")
print(f"Final Even Sum: {even_result}")
```

# 18. 
Python Program to Find Factorial of Number Using Recursion

```python
def factorial_recursive(n):
    if n == 1 or n == 0:
        return 1
    else:
        return n * factorial_recursive(n - 1)

num = 5
result = factorial_recursive(num)
print(f"The factorial of {num} is {result}")
```
